### Постановка задачи 1.13
Найти экстремум целевой функции:
$$ f(x) = (x_1^6 - x_2^2)^2 + 5x_2^2 - 10x_1 \to \text{extr} $$
при ограничениях-неравенствах:
1. $ g_1(x) = x_1^2 + 4x_2^2 - 8 \le 0 $
2. $ g_2(x) = -x_1 \le 0 $

$g_2(x) = x_1 \ge 0 $

## Шаг 1. Составление функции Лагранжа

Поскольку у нас задача только с ограничениями-неравенствами, составим классическую функцию Лагранжа (полагая $\lambda_0 = 1$, так как вырожденный случай $\lambda_0 = 0$ решений не имеет):
$$ L(x, \lambda) = f(x) + \lambda_1 g_1(x) + \lambda_2 g_2(x) $$
$$ L(x, \lambda) = (x_1^6 - x_2^2)^2 + 5x_2^2 - 10x_1 + \lambda_1(x_1^2 + 4x_2^2 - 8) + \lambda_2(-x_1) $$

## Шаг 2. Записать необходимые условия экстремума первого порядка

Найдем частные производные функции Лагранжа:
$$ \frac{\partial L}{\partial x_1} = 2(x_1^6 - x_2^2) \cdot 6x_1^5 - 10 + 2\lambda_1 x_1 - \lambda_2 = 12x_1^5(x_1^6 - x_2^2) - 10 + 2\lambda_1 x_1 - \lambda_2 $$
$$ \frac{\partial L}{\partial x_2} = 2(x_1^6 - x_2^2) \cdot (-2x_2) + 10x_2 + 8\lambda_1 x_2 = -4x_2(x_1^6 - x_2^2) + 10x_2 + 8\lambda_1 x_2 $$

Запишем систему необходимых условий (условия стационарности, дополняющей нежесткости и допустимости):

А)
$$12x_1^5(x_1^6 - x_2^2) - 10 + 2\lambda_1 x_1 - \lambda_2 = 0$$
$$2x_2 \big[ -2(x_1^6 - x_2^2) + 5 + 4\lambda_1 \big] = 0$$
Б)
$$x_1^2 + 4x_2^2 - 8 \le 0$$
$$x_1 \ge 0$$
Г)
$$\lambda_1 (x_1^2 + 4x_2^2 - 8) = 0$$
$$\lambda_2 (-x_1) = 0$$


## Шаг 3. Решить систему уравнений для двух случаев

#### Случай 1: $x_1 = 0$
Если $x_1 = 0$ то $\lambda_2$ может быть любым.
Подставим $x_1 = 0$ в А:
$$12(0)^5(0 - x_2^2) - 10 + 2\lambda_1(0) - \lambda_2 = 0 $$
$$\lambda_2 = -10$$
$$2x_2 \big( 2x_2^2 + 5 + 4\lambda_1 \big) = 0$$
Отсюда либо $x_2 = 0$, либо $2x_2^2 + 5 + 4\lambda_1 = 0$.

* $x_2 = 0$.
    Тогда $g_1(0,0) = -8 < 0$ (ограничение пассивно), следовательно $\lambda_1 = 0$.
    Получаем точку **$M_1(0, 0)$** с множителями $\lambda_1 = 0$, $\lambda_2 = -10$.

* $x_2 \neq 0$, $2x_2^2 + 5 + 4\lambda_1 = 0$.
    Либо $\lambda_1 = 0$, либо $0^2 + 4x_2^2 - 8 = 0$.
    Если $\lambda_1 = 0$, то $2x_2^2 + 5 = 0$ - нет действительных решений.
    Значит $$4x_2^2 = 8 $$ $$ x_2^2 = 2 $$ $$ x_2 = \pm\sqrt{2}$$
    Тогда $$2(2) + 5 + 4\lambda_1 = 0 $$
    $$ 4\lambda_1 = -9 $$
    $$ \lambda_1 = -2.25$$
    Получаем две точки $M_2(0, \sqrt{2})$ и $M_3(0, -\sqrt{2})$ с $\lambda_1 = -2.25$, $\lambda_2 = -10$.

#### Случай 2: $x_1 > 0$
Раз $x_1 \neq 0$ то $\lambda_2 = 0$.
Уравнение A.1 принимает вид: $12x_1^5(x_1^6 - x_2^2) - 10 + 2\lambda_1 x_1 = 0$.
Из уравнения A.2 либо $x_2 = 0$ либо скобка равна нулю.

* $x_2 = 0$.
    Уравнение A.1 становится: $12x_1^{11} - 10 + 2\lambda_1 x_1 = 0$.
    Из Г.1: либо $\lambda_1 = 0$ либо $x_1^2 = 8$.
    - Если $\lambda_1 = 0$: $12x_1^{11} = 10 $
      $$ x_1 = \sqrt[11]{\frac{5}{6}} \approx 0.983$$
      Проверяем: $(0.983)^2 - 8 < 0$ .
      Получаем точку $M_4\left( \sqrt[11]{\frac{5}{6}}, 0 \right)$ с $\lambda_1 = 0$, $\lambda_2 = 0$.
    - Если $x_1^2 = 8 \implies x_1 = 2\sqrt{2}$ (т.к. $x_1 > 0$).
      Найдём $\lambda_1$: $12(2\sqrt{2})^{11} - 10 + 2\lambda_1(2\sqrt{2}) = 0 $
      $$ 12 \cdot 65536\sqrt{2} - 10 + 4\sqrt{2}\lambda_1 = 0$$.
      Отсюда $\lambda_1 = \frac{5\sqrt{2}}{4} - 196608 < 0$.
      Получаем точку **$M_5(2\sqrt{2}, 0)$** с $\lambda_1 \approx -196608$, $\lambda_2 = 0$.

* $x_2 \neq 0$ и $-2(x_1^6 - x_2^2) + 5 + 4\lambda_1 = 0$.
    https://www.wolframalpha.com/input?i=12x_1%5E5%28x_1%5E6+-+%282+-+0.25x_1%5E2%29+%29+-+10+%2B+2x_1+%280.5%28x_1%5E6+-+%282+-+0.25x_1%5E2%29%29+-+1.25%29+%3D+0
В этих точках $\lambda_1 < 0$ однако второй дифференциал $d^2L > 0$ что противоречит как условиям максимума, так и минимума.+

## Шаг 4. Проверка достаточных условий экстремума

Используем критерии второго порядка для обобщенной функции Лагранжа. Найдем вторые производные:
$L_{x_1 x_1} = 132x_1^{10} - 60x_1^4 x_2^2 + 2\lambda_1$
$L_{x_1 x_2} = -24x_1^5 x_2$
$L_{x_2 x_2} = -4x_1^6 + 12x_2^2 + 10 + 8\lambda_1$

*   Точка $M_1(0, 0)$ ($\lambda_1 = 0, \lambda_2 = -10$)
    Условие для минимума нарушено ($\lambda_2 < 0$). Проверим на максимум ($d^2L \le 0$).
    На касательном подпространстве $dx_1 = 0$ имеем $d^2L = L_{x_2 x_2}dx_2^2 = 10dx_2^2 > 0$. Знак $d^2L$ не соответствует требованию для максимума. $M_1$ - экстремума нет.

*   Точки $M_2(0, \sqrt{2})$ и $M_3(0, -\sqrt{2})$ ($\lambda_1 = -2.25, \lambda_2 = -10$)
    Для обеих точек активны 2 ограничения ($g_1=0, g_2=0$).
    Количество активных ограничений $l = 2$ совпадает с количеством переменных $n = 2$.
    Следовательно $M_2$ и $M_3$ - точки строгого условного локального максимума.

*   Точка $M_4\left( \sqrt[11]{5/6}, 0 \right)$ ($\lambda_1 = 0, \lambda_2 = 0$)
    Точка находится строго внутри области. Оба множителя $\ge 0$ (подходит для минимума).
    Вычислим матрицу Гессе:
    $L_{x_1 x_1} = 132(5/6)^{10/11} > 0$
    $L_{x_1 x_2} = 0$
    $L_{x_2 x_2} = -4(5/6)^{6/11} + 10 \approx 6.3 > 0$
    Второй дифференциал $d^2L > 0$ строго положителен для любых ненулевых $dx$.
    Следовательно $M_4$ - точка строгого локального минимума.

*   Точка $M_5(2\sqrt{2}, 0)$ ($\lambda_1 \approx -196608 < 0, \lambda_2 = 0$)
    Условие знака $\lambda_1 \le 0$ выполнено для максимума. Активно первое ограничение ($l=1$).
    $dg_1 = 4\sqrt{2}dx_1 + 0dx_2 = 0 \implies dx_1 = 0$.
    Проверим $d^2L$ при $dx_1 = 0$ и $dx_2 \neq 0$:
    $d^2L = L_{x_2 x_2} dx_2^2 = (-4(2\sqrt{2})^6 + 12(0)^2 + 10 + 8 \cdot \lambda_1) dx_2^2$.
    Так как $\lambda_1$ - огромное отрицательное число, то $L_{x_2 x_2} < 0 \implies d^2L < 0$.
    $M_5$ - точка строгого условного локального максимума.


## Шаг 5. Вычисление значений функции в точках экстремума

1. **Глобальный (и локальный) минимум:**
   $$f(M_4) = f\left(\sqrt[11]{\frac{5}{6}}, 0\right) = \left(\left(\frac{5}{6}\right)^{\frac{6}{11}} - 0\right)^2 + 0 - 10\left(\frac{5}{6}\right)^{\frac{1}{11}} = \left(\frac{5}{6}\right)^{\frac{12}{11}} - 10\left(\frac{5}{6}\right)^{\frac{1}{11}}$$
   Вынося общий множитель:
   $$f(M_4) = \left(\frac{5}{6}\right)^{\frac{1}{11}} \cdot \left( \frac{5}{6} - 10 \right) = - \frac{55}{6} \left(\frac{5}{6}\right)^{\frac{1}{11}} \approx -9.014$$

2. **Локальные максимумы (на пересечении ограничений):**
   $$f(M_2) = f(M_3) = f(0, \pm\sqrt{2}) = (0^6 - 2)^2 + 5(2) - 10(0) = (-2)^2 + 10 = 14$$

3. **Глобальный (и локальный) максимум:**
   $$f(M_5) = f(2\sqrt{2}, 0) = ((2\sqrt{2})^6 - 0)^2 + 5(0) - 10(2\sqrt{2})$$
   $$(2\sqrt{2})^6 = (\sqrt{8})^6 = 8^3 = 512$$
   $$f(M_5) = 512^2 - 20\sqrt{2} = 262144 - 20\sqrt{2} \approx 262115.7$$

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import minimize

# 1. Определение целевой функции
def f(x):
    x1, x2 = x
    return (x1**6 - x2**2)**2 + 5 * x2**2 - 10 * x1

# Для поиска максимума будем минимизировать функцию, взятую со знаком минус
def minus_f(x):
    return -f(x)

# 2. Определение ограничений
# g1(x) = x1^2 + 4*x2^2 - 8 <= 0 
# 8 - x1^2 - 4*x2^2 >= 0
cons = ({'type': 'ineq', 'fun': lambda x: 8 - x[0]**2 - 4*x[1]**2})

# g2(x) = -x1 <= 0  ==>  x1 >= 0
bounds = ((0, None), (None, None))  # x1 in [0, +inf), x2 in (-inf, +inf)

# 3. Численный поиск экстремумов

# Стартуем из точки (1, 0), так как аналитически корень x1 ~ 0.983
res_min = minimize(f, x0=[1.0, 0.0], bounds=bounds, constraints=cons)

# Стартуем из разных областей допустимого множества, чтобы найти все локальные максимумы
res_max1 = minimize(minus_f, x0=[0.0, 1.4], bounds=bounds, constraints=cons)
res_max2 = minimize(minus_f, x0=[0.0, -1.4], bounds=bounds, constraints=cons)
res_max3 = minimize(minus_f, x0=[2.8, 0.0], bounds=bounds, constraints=cons)

if res_min.success:
    print(f"Локальный (и глобальный) минимум: x = [{res_min.x[0]:.4f}, {res_min.x[1]:.4f}], f(x) = {res_min.fun:.4f}")

if res_max1.success:
    print(f"Локальный максимум 1: x = [{res_max1.x[0]:.4f}, {res_max1.x[1]:.4f}], f(x) = {-res_max1.fun:.4f}")

if res_max2.success:
    print(f"Локальный максимум 2: x = [{res_max2.x[0]:.4f}, {res_max2.x[1]:.4f}], f(x) = {-res_max2.fun:.4f}")

if res_max3.success:
    print(f"Глобальный максимум: x = [{res_max3.x[0]:.4f}, {res_max3.x[1]:.4f}], f(x) = {-res_max3.fun:.4f}")

plt.figure(figsize=(10, 8))

x1_vals = np.linspace(-0.5, 3.2, 400)
x2_vals = np.linspace(-1.8, 1.8, 400)
X1, X2 = np.meshgrid(x1_vals, x2_vals)
Z = f([X1, X2])

levels = np.linspace(-10, 100, 30)
cp = plt.contourf(X1, X2, Z, levels=levels, extend='max', alpha=0.8)
plt.colorbar(cp, label='Значение f(x)')

# Отрисовка границы ограничения x1^2 + 4x2^2 = 8
theta = np.linspace(0, 2*np.pi, 200)
ellipse_x1 = np.sqrt(8) * np.cos(theta)
ellipse_x2 = np.sqrt(2) * np.sin(theta)
plt.plot(ellipse_x1, ellipse_x2, 'r--', linewidth=2, label='Ограничение $x_1^2 + 4x_2^2 - 8 \leq 0$')

# Ограничение x1 >= 0
plt.axvline(x=0, color='blue', linestyle='--', linewidth=2, label='Ограничение $x_1 \geq 0$')

# Отмечаем найденные точки на графике
plt.plot(res_min.x[0], res_min.x[1], 'w*', markersize=12, markeredgecolor='k', label='Минимум $M_4$')
plt.plot(res_max1.x[0], res_max1.x[1], 'r^', markersize=12, markeredgecolor='k', label='Максимум $M_2, M_3$')
plt.plot(res_max2.x[0], res_max2.x[1], 'r^', markersize=12, markeredgecolor='k')
plt.plot(res_max3.x[0], res_max3.x[1], 'rs', markersize=12, markeredgecolor='k', label='Максимум $M_5$')

# Настройки графика
plt.title('Линии уровня целевой функции и ограничения')
plt.xlabel('$x_1$')
plt.ylabel('$x_2$')
plt.xlim([-0.5, 3.2])
plt.ylim([-1.8, 1.8])
plt.grid(True, linestyle=':', alpha=0.7)
plt.legend(loc='upper right')

plt.show()
```
