# 1. Текстовое описание набора данных
В качестве набора данных мы будем использовать набор данных по определению вида вина по его составу и характеритикам - https://archive.ics.uci.edu/ml/machine-learning-databases/wine/wine.data

Датасет включает в себя 13 параметров представленных числами:

* Alcohol - содержание алкоголя
* Malic acid - содержание яблочной кислоты
* Ash - минеральный остаток оставшийся в ходе сжигания остатка после испарения вина
* Alcalinity of ash - щелочность минерального остатка
* Magnesium - содержание магния
* Total phenols - общее количество фенолов
* Flavanoids - содержание флаваноидов
* Nonflavanoid phenols - содержание нефлаваноидовыъ фенолов
* Proanthocyanins - содержание Проантоцианидины
* Color intensity - интенсивность цвета
* Hue - отеннок
* OD280/OD315 of diluted wines - Оптическая плотность при 280 нм и 315 нм разбавленных вин
* Proline - содержание пролина


```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
%matplotlib inline
sns.set(style="ticks")
```

```python
data = pd.read_csv('https://archive.ics.uci.edu/ml/machine-learning-databases/wine/wine.data', sep=",")
data.columns = ['Class', 'Alcohol', 'Malic acid', 'Ash', 'Alcalinity of ash', 'Magnesium', 'Total phenols', 'Flavanoids', 'Nonflavanoid phenols', 'Proanthocyanins', 'Color intensity', 'Hue', 'OD280/OD315 of diluted wines', 'Proline']
```

# 4. Информация о корреляции признаков.

```python
data.corr()
```

На основе корреляционной матрицы можно сделать следующие выводы:

* Тип вина больше всего корредилует с содержанием флаваноидов (-0.84), коэффициентом преломления (-0.78) и общим содержанием фенолов (-0.71)
* Содержание флаваноидов сильно коррелирует (0.86) с общий содержанием фенолов, между ними имеется зависимость (флаваноиды входят в фенолы). Из двух лучше оставить содержание флаваноидов так как этот признак коррелирует с целевым больше.
* Целевой признак отчасти коррелирует с пролине (-0.63) и щелочностью минерального остатка (0.51)
* Содержание минерального остатка крайне мало коррелирует с целевым признаком (0.04) его стоит исключить
* Содержание магния (0.19) и интенсивность цвета (0.26) мало коррелируют с целевым признаков, они могут негативно повлиять на модель

```python
data.corr(method='pearson')
```

```python
data.corr(method='kendall')
```

```python
data.corr(method='spearman')
```

Все три метода показывают очень похожие результаты

```python
fig, ax = plt.subplots(1, 1, sharex='col', sharey='row', figsize=(20,10))
sns.heatmap(data.corr(), annot=True, fmt='.3f')
```

```python
fig, ax = plt.subplots(1, 3, sharex='col', sharey='row', figsize=(30,5))
sns.heatmap(data.corr(method='pearson'), ax=ax[0], annot=True, fmt='.1f')
sns.heatmap(data.corr(method='kendall'), ax=ax[1], annot=True, fmt='.1f')
sns.heatmap(data.corr(method='spearman'), ax=ax[2], annot=True, fmt='.1f')
fig.suptitle('Корреляционные матрицы, построенные различными методами')
ax[0].title.set_text('Pearson')
ax[1].title.set_text('Kendall')
ax[2].title.set_text('Spearman')
```
